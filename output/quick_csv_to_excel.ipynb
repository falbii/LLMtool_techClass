# Quick CSV to Excel
This notebook picks one CSV file from this folder (`output/`) and saves it as an Excel file (`.xlsx`).

```python
from pathlib import Path
import pandas as pd

folder = Path.cwd()  # run from output/
csv_files = sorted(folder.glob('*.csv'))

if not csv_files:
    raise FileNotFoundError('No CSV files found in this folder.')

print('Available CSV files:')
for i, file in enumerate(csv_files, start=1):
    print(f'{i}. {file.name}')

choice = int(input('Select CSV number: ')) - 1
if choice < 0 or choice >= len(csv_files):
    raise ValueError('Invalid selection.')

csv_file = csv_files[choice]
excel_name = csv_file.stem
excel_file = folder / f'{excel_name}.xlsx'

df = pd.read_csv(csv_file)
df.to_excel(excel_file, index=False)

print(f'Converted!')
```
