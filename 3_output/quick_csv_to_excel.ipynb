# Quick CSV to Excel
This notebook picks one CSV file from this folder or its subfolders (e.g. `2_csv_classification/`) and saves it as an Excel file (`.xlsx`) next to the source CSV.

```python
from pathlib import Path
import pandas as pd

folder = Path.cwd()  # run from 3_output/
csv_files = sorted(folder.rglob('*.csv'))  # also searches subfolders like 2_csv_classification/

if not csv_files:
    raise FileNotFoundError('No CSV files found in this folder or its subfolders.')

print('Available CSV files:')
for i, file in enumerate(csv_files, start=1):
    print(f'{i}. {file.relative_to(folder)}')

choice = int(input('Select CSV number: ')) - 1
if choice < 0 or choice >= len(csv_files):
    raise ValueError('Invalid selection.')

csv_file = csv_files[choice]
excel_file = csv_file.with_suffix('.xlsx')  # saved next to the source CSV

df = pd.read_csv(csv_file)
df.to_excel(excel_file, index=False)

print(f'Converted! -> {excel_file.relative_to(folder)}')
```
